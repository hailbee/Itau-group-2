```python
import pandas as pd
import itertools

# read typosquats.csv as df
df = pd.read_csv("typosquats.csv")

# remove all rows where 'classification' == 'other'
df = df[df["classification"] != "other"]

# add a column called label and set it to 1 for all rows
df["label"] = 1

# rename columns
df = df.rename(columns={
    "malicious_package": "fraudulent_name",
    "target_package": "real_name"
})

# keep only the required columns
df = df[["fraudulent_name", "real_name", "label"]]

# get all unique names in target_package (now real_name)
unique_names = df["real_name"].unique()

# generate all unique pairs of those names
pairs = list(itertools.combinations(unique_names, 2))

# create dataframe of negative examples
neg_df = pd.DataFrame(pairs, columns=["fraudulent_name", "real_name"])
neg_df["label"] = 0

# combine with original dataframe
df_final = pd.concat([df, neg_df], ignore_index=True)

df_final.head()

# export to csv called typosquat_ref.csv
df_final.to_csv("typosquat_ref.csv", index=False)

# export to parquet called typosquat_ref.parquet
df_final.to_parquet("typosquat_ref.parquet", index=False)
```
